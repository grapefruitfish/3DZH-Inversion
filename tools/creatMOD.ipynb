```python
import numpy as np

def create_mod_file(depths, vs_values, lon_grid, lat_grid, output_filename="MOD"):
    """
    创建 MOD 文件
    
    格式：
    - 第1行：depths
    - 后续：对每个 depth，逐个 longitude 写一行
      每一行是该 depth、该 longitude 下，不同 latitude 的 Vs 值
    
    也就是顺序：
    depth -> longitude -> latitude
    """

    if len(depths) != len(vs_values):
        raise ValueError("depths 和 vs_values 长度必须一致")

    mod_content = []

    # 第1行：深度
    mod_content.append(" ".join(f"{d:.1f}" for d in depths))

    # 后续：depth -> lon -> lat
    for idep, vs in enumerate(vs_values):
        for ilon in range(len(lon_grid)):
            # 对于1D初始模型，该深度下所有经纬网格值都相同
            row = [vs] * len(lat_grid)
            mod_content.append(" ".join(f"{v:.3f}" for v in row))

    with open(output_filename, "w") as f:
        f.write("\n".join(mod_content))

    print(f"MOD 文件已生成: {output_filename}")
    print("文件结构：")
    print(f"- 第1行: {len(depths)} 个深度值")
    print(f"- 后续共 {len(depths)} × {len(lon_grid)} = {len(depths) * len(lon_grid)} 行")
    print(f"- 每行 {len(lat_grid)} 个速度值（对应不同纬度）")

# 示例数据
depths = np.array([0.0, 2.5, 5.0, 7.5, 10.0, 15.0, 20.0, 25.0, 30.0, 35.0, 40.0, 50.0, 60.0, 70.0])
vs_values = np.array([3.078, 3.131, 3.266, 3.407, 3.469, 3.487, 3.488, 3.506, 3.555, 3.782, 3.935, 4.157, 4.345, 4.394])
lon_grid = np.arange(96.0,109.0,0.2)
lat_grid = np.arange(20,31,0.2)
create_mod_file(depths, vs_values, lon_grid, lat_grid, "MOD")
print(len(lat_grid), len(lon_grid),len(depths))
```
